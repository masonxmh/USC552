#### The iterrows() Method

copy the code above to the following cell and add code for plotting curve $\left(x, y1\right)$, $\left(x, y2\right)$ and $\left(x, y3\right)$.

````python
import numpy as np
import matplotlib.pyplot as plt


x = np.linspace(-5, 5, num=20)
y1 = np.array([j for j in x])
y2 = np.array([j ** 2 for j in x])
y3 = np.array([j ** 3 for j in x])



fig, axs = plt.subplots(3, figsize=(6,18))

axs[0].plot(x,y1, marker ='o', color='blue', label=r'y1=$x$')
axs[0].set_title("y1 vs. x", fontsize=15)

axs[1].plot(x,y2, marker='^', color='red', label=r'y2=$x^2$')
axs[1].set_title("y2 vs. x", fontsize=15)

axs[2].plot(x,y3, marker='d', color='orange',label=r'y3=$x^3$')
axs[2].set_title("y3 vs. x", fontsize=15)


for ax in axs:
#     ax.set(xlabel='x-label', ylabel='y-label')
    ax.set_xlabel('x-label',fontsize=15)
    ax.set_ylabel('y-label',fontsize=15)
    ax.grid(color='g', linestyle='--', linewidth=1)
    ax.legend(loc="lower right",prop={"size":15})
plt.show()
````
